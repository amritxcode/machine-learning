```python
## Polimorphism with abstract base class
from abc import ABC, abstractmethod

## Define an abstract class
class Vehicle(ABC):
  @abstractmethod
  def start_engine(self):
    pass
## Derived Class 1
class Car(Vehicle):
  def start_engine(self):
    print("Car engine started")
## Derived Class 2
class Bike(Vehicle):
  def start_engine(self):
    print("Bike engine started")
## Derived Class 3
class Bus(Vehicle):
  def start_engine(self):
    print("Bus engine started")

car = Car()
bike = Bike()
bus = Bus()

car.start_engine()
bike.start_engine()
bus.start_engine()
```
