```python
# Read full file
with open('python_file_practice.txt') as file:
  content = file.read()
  print(content)
```

```python
## Read a file line by line
with open('python_file_practice.txt') as file:
  for line in file:
    print(line.strip())
```

```python
### Writing a list of lines to a file
lines = ['line 1', 'line 2', 'line 3']
with open('python_file_practice.txt','a') as file:
  file.writelines(lines)
```
