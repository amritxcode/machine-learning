```python
## Matplotlib is a powerful plotting library for Python that enables the creation of static, animated, and interactive visualization.
```

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [1,4,9,16,25]

## Create a line plot
plt.plot(x,y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Basic Line Plot')
plt.show()
```

```python
x = [1,2,3,4,5]
y = [1,4,9,16,25]

## Create a customized line plot
plt.plot(x,y,color = 'red',linestyle = '--',marker = 'o', linewidth = 1.5, markersize=9)
plt.grid(True)
```

```python
## Multiple Plots
##V Sample data
x = x = [1,2,3,4,5]
y = [1,4,9,16,25]
y1 = [1,2,3,4,5]
y2 = [1,4,9,16,25]

## Create a line plot

plt.figure(figsize=(9,5))

plt.subplot(2,2,1)
plt.plot(x,y,color = 'green')
plt.title('Plot 1')

plt.subplot(2,2,2)
plt.plot(y1,x,color = 'blue')
plt.title('Plot 2')

plt.subplot(2,2,3)
plt.plot(y1,x,color = 'red')
plt.title('Plot 3')


plt.subplot(2,2,4)
plt.plot(x,y2,color = 'yellow')
plt.title('Plot 4')
```

```python
## Bar plot
categories= ['A', 'B','C', 'D', 'E']
values = [5,7,3,8,6]

#create a bar plot
plt.bar(categories,values,color = 'purple')
```
