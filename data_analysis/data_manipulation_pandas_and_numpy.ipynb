```python
print(df['Performance_Score'].min(),
df['Performance_Score'].max()
)
```

```python
df['Performance_Score'] = df['Performance_Score'].fillna(df['Performance_Score'].mean())
print(df.isnull().sum().sort_values(ascending=False))
df
```

```python
print(df['Salary'].min(),
df['Salary'].max()
)
print('\n',df['Salary'].mean(),
df['Salary'].median()
)
```

```python
df['Salary']= df['Salary'].fillna(df['Salary'].median())
print(df.isnull().sum().sort_values(ascending=False))
df
```

```python
df['Join_Date']=df['Join_Date'].fillna(df['Join_Date'].median())
print(df.isnull().sum().sort_values(ascending=False))
df
```

```python
df['Department']=df['Department'].fillna(df['Department'].mode()[0])
print(df.isnull().sum().sort_values(ascending=False))
df
```

```python
df.dtypes
```

```python
df['Remote']= df['Remote'].fillna(df['Remote'].mode()[0])
df['Remote'] = df['Remote'].astype(bool)
print(df.isnull().sum().sort_values(ascending=False))
df
```
