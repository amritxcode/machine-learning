```python
# The most useful plot for ML — correlation heatmap
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap='coolwarm',
            fmt='.2f')
plt.title('Correlation Heatmap — Titanic Dataset')
plt.show()
```
