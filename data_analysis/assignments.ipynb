```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## Assignment 1 Load any dataset from a URL
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)
df.head()
```

```python
## Assignment 2 Plot a histogram of any numerical column
plt.figure(figsize = (10,6))
plt.hist(df['Age'], bins = 30, edgecolor = 'black')
plt.title('Distribution By Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
## Assignment 3: Create a correlation heatmap
plt.figure(figsize = (10,6))
sns.heatmap(df.corr(numeric_only=True), annot = True, cmap = 'rocket')
```

```python
sns.countplot(data = df, x = 'Pclass', hue = 'Survived')
plt.title('Survived By Age')
plt.show()
```

```python
# Assignment 4
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv'
df = pd.read_csv(url)
df.head()
```

```python
# Assignment 5
plt.figure(figsize= (10,6))
plt.hist(df['total_bill'], bins = 30, edgecolor = 'black')
plt.title("Distribution By Total Bill")
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

```python
# plt.figure(figsize= (10,6))
sns.countplot(data = df, x = 'day', hue = 'sex')
plt.title('Count By Day')
plt.show()
```

```python
sns.scatterplot(data = df, x = 'total_bill', y = 'tip', hue = 'sex')
plt.title('Total Bill vs Tip')
plt.show()
```
