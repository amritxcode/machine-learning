```python
# Plot 1 — Age distribution
plt.figure(figsize=(10,6))
plt.hist(df['Age'].dropna(), bins=30, color='steelblue', edgecolor='black')
plt.title('Age Distribution of Titanic Passengers')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()

# Plot 2 — Survival by class
plt.figure(figsize=(10,6))
plt.bar(df['Pclass'].value_counts().index,
        df['Pclass'].value_counts().values,
        color=['gold','silver','brown'])
plt.title('Passengers per Class')
plt.xlabel('Class')
plt.ylabel('Count')
plt.show()
```
