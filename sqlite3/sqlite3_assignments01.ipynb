```python
import sqlite3
```

```python
conn = sqlite3.connect('test.db')
cursor = conn.cursor()
```

```python
cursor.execute("DROP TABLE IF EXISTS inventory")
conn.commit()
# Then run your CREATE TABLE script again
```

```python
cursor.execute('''
create table if not exists inventory(
item_id integer primary key,
item_name text unique,
quantity float,
price real
)
''')
conn.commit()
```

```python
cursor.execute('insert or ignore into inventory(item_name, quantity, price) values (?, ?, ?)', ('Rice', 5, 250))
cursor.execute('insert or ignore into inventory(item_name, quantity, price) values (?, ?, ?)', ('Dal', 5, 500))
cursor.execute('insert or ignore into inventory(item_name, quantity, price) values (?, ?, ?)', ('Sugar', 2, 100))
conn.commit()
```

```python
cursor.execute('Select * FROM inventory where quantity < 3')
print(cursor.fetchall())
```

```python
cursor.execute("update inventory set price = 100 where item_name = 'Sugar'")
conn.commit()
```

```python
cursor.execute('Select * from inventory')
print(cursor.fetchall())
```

```python
cursor.execute("update inventory set price = price + (price*0.1) where quantity < 3")
conn.commit()
cursor.execute('select * from inventory')
print(cursor.fetchall())
```
