```python
import sqlite3
conn = sqlite3.connect('tasks.db')
cursor = conn.cursor()
```

```python
cursor.execute ('''
create table if not exists task (
id integer primary key,
title text not null,
priority text check(priority in ('High', 'Medium', 'Low')),
completed integer check(completed in (0, 1))
)
''')
conn.commit()
```

```python
def add_task(title, priority):
  try:
    cursor.execute("insert or ignore into task (title, priority, completed) values (?,?,?)",(title, priority, 0))
    conn.commit()
    print(f"Successfully added task: {title}")
  except sqlite3.Error as e:
    print(f"Dataase error: {e}")

def get_all_tasks():
  try:
    cursor.execute("select * from task")
    rows = cursor.fetchall()
    for row in rows:
      print(row)
  except sqlite3.Error as e:
    print(f"Database error: {e}")

def complete_task(id):
  try:
    cursor.execute("update task set completed = 1 where id = ?", (id,))
    conn.commit()
    print(f"Successfully marked task ID {id} as complete")
  except sqlite3.Error as e:
    print(f"Database error: {e}")

def delete_task(id):
  try:
    cursor.execute("delete from task where id = ?", (id,))
    conn.commit()
    print(f"Successfully delete task ID {id}")
  except sqlite3.Error as e:
    print(f'Database error: {e}')
```

```python
add_task('Hello', 'Low')
```
