```python
import sqlite3
```

```python
# 1. CONNECT - creates database file if it doesn't exist
conn = sqlite3.connect('test.db')
cursor = conn.cursor()
```

```python
# 2. Create - make a table
cursor.execute('''
Create Table if not exists students(
id integer primary key,
name text not null,
age integer,
grade text
)
''')
conn.commit()
```

```python
# 3 insert (create in CRUD) - add data
cursor.execute('Insert into students( name, age, grade) VALUES (?, ?, ?)',('Amrit', 22, 'A'))
cursor.execute('Insert into students( name, age, grade) VALUES (?, ?, ?)',('Appu', 22, 'A'))
conn.commit()
```

```python
# 4. select (Read in CRUD) - get data
cursor.execute('select * from students')
rows = cursor.fetchall()
for row in rows:
  print(rows)
```

```python
# 5.  UPDATE — change existing data
cursor.execute('update students set grade = ? where name = ?', ('A', 'Amrit'))
conn.commit()
```

```python
cursor.execute('DELETE FROM students WHERE name IN (?, ?)', ('Appu', 'Amrit'))
conn.commit()
```
