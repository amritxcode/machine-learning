```python
import logging

logging.basicConfig(
    level = logging.DEBUG,
    format = '%(asctime)s - %(levelname)s - %(message)s',
    handlers = [
        logging.FileHandler('app.log'),
        logging.StreamHandler(),
    ],
    force = True
)
```

```python
def process_order(item, stock):
  logging.info(f'Starting order processing of item: {item.title()}.')

  if stock == 0:
    logging.warning(f'{item.title()} is out of stock.')
    return False
  else:
    logging.info(f'Order placed successfully.')
    return True

process_order('rice', 0)
```
