```python
import logging

logging.basicConfig(
    level = logging.DEBUG,
    format = '%(asctime)s - %(levelname)s - %(message)s',
    handlers =[
        logging.FileHandler('app.log'),
        logging.StreamHandler(),
    ],
    force = True
)
```

```python
def check_age(age):
  logging.info(f'Starting age verification for input: {age}')

  if age < 18:
    logging.warning(f'Underage user blocked. Age provided: {age}')
    return False
  else:
    logging.info(f'Age verification passed. Age provided: {age}')
    return True

check_age(20)
check_age(15)
```
