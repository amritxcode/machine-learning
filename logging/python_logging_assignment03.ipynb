```python
import logging

logging.basicConfig(
    level = logging.DEBUG,
    format = '%(asctime)s - %(levelname)s - %(message)s',
    datefmt = '%d-%b-%y %H:%M:%S',
    handlers = [
        logging.FileHandler('app.log'),
        logging.StreamHandler(),
    ],
    force = True
)
```

```python
def divide_num(a, b):
  logging.info(f'Initializing division of {a} and {b}.')

  if b == 0:
    logging.warning('Divisor is zero. Division cannot proceed safely.')
    return None
  try:
    logging.info(f'Division of {a} and {b} is successful.')
    return a/b
  except Exception as e:
    logging.error(f'Unexpected calculation failure: {e}')
    return None

divide_num(10, 5)
divide_num(10, 0)
```
