```python
import logging

logging.basicConfig(
    level = logging.DEBUG,
    format = '%(asctime)s - %(levelname)s - %(message)s',
    handlers = [
        logging.FileHandler('app.log'),
        logging.StreamHandler(),
    ],
    force = True
)
```

```python
def withdraw_money(balance, amount):
  logging.info(f'Attempting withdrawal of {amount} from balance {balance}')

  if amount > balance:
    logging.warning(
        f'Withdrawal denied: Requested {amount} exceeds balance {balance}'
    )
    return balance

  try:
    new_balance = balance - amount
    logging.info(f'Transaction successful. Remaining balance: {new_balance}')
    return new_balance

  except Exception as e:
    logging.error(f'An error occurred: {e}')
    return balance

balance = 1000
balance = withdraw_money(balance, 500)
balance = withdraw_money(balance, 1500)
balance = withdraw_money(balance, 200)
```
